```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset=pd.read_csv("Algerian.csv")
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
### data cleaning
### mising values
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122,'Region']=0
dataset.loc[122:,'Region']=1
df=dataset
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
pd.read_csv("sidi-Bel.csv")
```

```python
### removing null values
df.dropna().reset_index(drop=True)
df.head()
```

```python
df.isnull().sum()
```

```python
### remove the 122 row
df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## fix spaces in column name
df.columns=df.columns.str.strip()
```

```python
df.columns
```

```python
## change the required columns as integer data type
df[['month','year','Temperature','RH','Ws']]=df[['month','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
### changing the opthers columns to float datatype
objects=[features for features in df.columns if df[features].dtype=='O']
##df[['rain','ffmc','DMC','DC','ISI','BUI','FWI']]=df[['rain','ffmc','DMC','DC','ISI','BUI','FWI']].astype(float)
```

```python
for i in objects:
  if i!='Classes':
    df[i]=df[i].astype(float)
```

```python
### let save the clean dataset
df.to_csv("cleaned_dataset.csv",index=False)
```
